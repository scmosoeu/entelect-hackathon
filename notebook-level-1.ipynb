```python
import ast
import networkx as nx
```

```python
with open('data/level_1.txt', 'r') as f:
    file_content = f.read()
```

```python
file_content
```

```python
data = ast.literal_eval(file_content)
```

```python
data
```

```python
type(data)
```

```python
data['adjacency_list']
```

```python
start_node = data['start']
end_note = data['end']
graph = data['adjacency_list']
```

```python
edges = []
for key, values in graph.items():
    for value in values:
        node = value['node']
        weight = value['weight']
        entry = (key, node, weight)
        edges.append(entry)
```

```python
edges
```

```python
# Create a weighted graph
G = nx.Graph()
G.add_weighted_edges_from(edges)
route = nx.dijkstra_path(G, source=start_node, target=end_note)
```
